```python
import numpy as np
import matplotlib.pyplot as plt

k = 8.99e9

# four charges at the corners of a square: (q, x, y)
charges = [(1, -2, -2), (1, 2, -2), (1, -2, 2), (1, 2, 2)]

x = np.linspace(-6, 6, 25)
y = np.linspace(-6, 6, 25)
X, Y = np.meshgrid(x, y)

Ex = np.zeros_like(X)
Ey = np.zeros_like(Y)

for q, x0, y0 in charges:
    dx, dy = X - x0, Y - y0
    r = np.hypot(dx, dy) + 1e-6   # avoid dividing by zero
    Ex += k * q * dx / r**3
    Ey += k * q * dy / r**3

mag = np.hypot(Ex, Ey)

fig, ax = plt.subplots()
ax.quiver(X, Y, Ex/mag, Ey/mag)   # unit-length arrows show direction only
for q, x0, y0 in charges:
    ax.plot(x0, y0, 'ro')
ax.set_aspect('equal')
plt.show()
```
